```python
from langchain_community.document_loaders import WebBaseLoader
```

```python
loader = WebBaseLoader("https://www.heum.ai/ko/apply-faq")
docs = loader.load()
```

```python
print(docs[0])
```

```python
from langchain_text_splitters import HTMLHeaderTextSplitter
```

```python
headers_to_split_on = [
    ("h1", "Header 1"),
    ("h2", "Header 2"),
    ("h3", "Header 3"),
]

html_splitter  = HTMLHeaderTextSplitter(headers_to_split_on)
```

```python
header_splits = html_splitter.split_text(docs[0].page_content)
```

```python
print(docs[0])
```

```python
print(header_splits[0].page_content)
```

```python
from langchain_huggingface.embeddings import HuggingFaceEndpointEmbeddings
```

```python
embeddings = HuggingFaceEndpointEmbeddings(client="http://127.0.0.1:8080/")
```

```python
embeddings
```

```python
from langchain_experimental.text_splitter import SemanticChunker
# from langchain_openai.embeddings import OpenAIEmbeddings

# 1. 임베딩 모델 준비
# embeddings = OpenAIEmbeddings()

# 2. 세만틱 청커 초기화 (분할 전략 선택 가능)
text_splitter = SemanticChunker(
    embeddings, 
    breakpoint_threshold_type="percentile" # 또는 "standard_deviation"
)
```

```python
header_splits[0].page_content
```

```python
text_splitter_standard = SemanticChunker(
    embeddings, 
    breakpoint_threshold_type="standard_deviation" # 또는 "standard_deviation"
)
```

```python
split_docs_standard = text_splitter_standard.create_documents([header_splits[0].page_content])
```

```python
# 3. 텍스트 분할 실행
# text = "사과는 과일입니다. 바나나도 과일입니다. 자동차는 이동수단입니다. 비행기도 탈것입니다."
split_docs = text_splitter.create_documents([header_splits[0].page_content])
```
