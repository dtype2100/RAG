## 1. Config 로드 확인

```python
from app.config import settings

print(f"임베딩 엔드포인트 : {settings.EMBEDDING_ENDPOINT}")
print(f"Intent 모델     : {settings.INTENT_MODEL} @ {settings.INTENT_BASE_URL}")
print(f"Schema 모델     : {settings.SCHEMA_MODEL}")
print(f"청크 크기        : {settings.CHUNK_SIZE} / 오버랩: {settings.CHUNK_OVERLAP}")
print(f"유사도 임계값    : {settings.SIMILARITY_THRESHOLD}")
print(f"타겟 스키마      : {settings.TARGET_SCHEMA}")
```

## 2. 임베딩 서버 헬스 체크

```python
import requests

url = settings.EMBEDDING_ENDPOINT.rstrip('/') + '/health'
try:
    resp = requests.get(url, timeout=5)
    print(f"상태 코드: {resp.status_code}")
    print(f"응답: {resp.text[:200]}")
except Exception as e:
    print(f"연결 실패: {e}")
    print("TEI 서버가 실행 중인지 확인하세요: make up-embedding")
```

## 3. 임베딩 생성 확인

```python
from app.embeddings import get_embeddings

embeddings = get_embeddings()
test_texts = ["인공지능", "머신러닝", "딥러닝"]
vectors = embeddings.embed_documents(test_texts)

print(f"임베딩 수       : {len(vectors)}")
print(f"벡터 차원       : {len(vectors[0])}")
print(f"첫 번째 벡터 앞 5개 값: {vectors[0][:5]}")
```
