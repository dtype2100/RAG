```python
# RAGAS 데이터셋 생성 프롬프트 예시
template = """당신은 RAGAS 평가용 데이터셋 생성기입니다.
아래 문맥만 근거로 한국어 샘플 2개를 생성하세요.
각 샘플은 반드시 아래 키를 포함해야 합니다.
- question: 사용자 질문
- ground_truth: 문맥 기반 정답
- contexts: 정답 근거 문장 배열(최소 1개)

출력 규칙:
1) 반드시 JSON 배열만 반환
2) 마크다운, 코드블록, 설명 문장 금지
3) 문맥에 없는 정보 추론/추가 금지

스키마:
[
  {{
    "question": "...",
    "ground_truth": "...",
    "contexts": ["..."]
  }},
  {{
    "question": "...",
    "ground_truth": "...",
    "contexts": ["..."]
  }}
]

문맥:
다음은 예시 문단입니다. 이 내용만 근거로 질문과 정답을 작성하세요.
"""
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama.llms import OllamaLLM


prompt = ChatPromptTemplate.from_template(template)

llm = OllamaLLM(
    model="gemma3-270m:latest",
    base_url="http://localhost:11434",
    temperature=0.2,
    ) 

chain = prompt | llm
```

```python
import time
```

```python
inputs = {
    "context": """
LangChain은 LLM 애플리케이션 개발을 위한 프레임워크다.
프롬프트 템플릿, 체인, 에이전트, 메모리, 리트리버 통합 기능을 제공한다.
Python과 JavaScript를 지원하며 RAG 파이프라인 구축에 자주 사용된다.
"""
}
# 2) 스트리밍 출력 (토큰/청크가 오는 즉시 출력)
start = time.perf_counter()
chunks = []
for chunk in chain.stream(inputs):
    print(chunk, end="", flush=True)
    chunks.append(chunk)
```

```python
elapsed = time.perf_counter() - start
full_text = "".join(chunks)
print("\n\n---")
print(f"elapsed: {elapsed:.2f}s")
print(f"chars: {len(full_text)}")
```
