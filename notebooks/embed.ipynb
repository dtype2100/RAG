```python
# https://docs.langchain.com/oss/python/integrations/splitters/recursive_text_splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
from qdrant_client import QdrantClient
from langchain_qdrant import QdrantVectorStore
from qdrant_client.models import Distance, VectorParams
client = QdrantClient(url="http://localhost:6333")
```

```python
with open("./md_pages/https___www_heum_ai_apply_faq.md") as f:
    md_file = f.read()
```

```python
print(md_file)
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    # Set a really small chunk size, just to show.
    chunk_size=512,
    chunk_overlap=50,
    length_function=len,
    is_separator_regex=False,
    separators=["*", "**", "#", "##", "###", "\n"]
)
```

```python
texts = text_splitter.create_documents([md_file])
```

```python
texts
```

```python
print(texts[5])
```

```python
# https://docs.langchain.com/oss/python/integrations/splitters/markdown_header_metadata_splitter
from langchain_text_splitters import MarkdownHeaderTextSplitter
```

```python
headers_to_split_on = [
    ("#", "Header 1"),
    ("##", "Header 2"),
    ("###", "Header 3"),
]

markdown_splitter = MarkdownHeaderTextSplitter(headers_to_split_on)
md_header_splits = markdown_splitter.split_text(md_file)
```

```python
print(md_header_splits[0])
```

```python
from langchain_huggingface.embeddings import HuggingFaceEndpointEmbeddings
embeddings = HuggingFaceEndpointEmbeddings(client="http://localhost:8080")
```

```python
vector_size = len(embeddings.embed_query(md_header_splits[0].page_content))
```

```python
vector_size
```

```python
if not client.collection_exists("test"):
    client.create_collection(
        collection_name="test",
        vectors_config=VectorParams(size=vector_size, distance=Distance.COSINE)
    )
```

```python

vector_store = QdrantVectorStore(
    client=client,
    collection_name="test",
    embedding=embeddings,
)
```

```python
vector_store.add_documents(md_header_splits)
```

```python
retriever = vector_store.as_retriever(
    search_type="similarity_score_threshold",
    search_kwargs={"score_threshold": 0.3},
)
```

```python
retriever.invoke("성과평가")
```

```python
await retriever.ainvoke('성과평가 ')
```

```python
from langchain_qdrant import FastEmbedSparse, RetrievalMode

sparse_embeddings = FastEmbedSparse(model_name="Qdrant/bm25")

qdrant = QdrantVectorStore.from_documents(
    docs,
    embedding=embeddings,
    sparse_embedding=sparse_embeddings,
    location=client,
    collection_name="my_documents",
    retrieval_mode=RetrievalMode.HYaBRID,
)

query = "What did the president say about Ketanji Brown Jackson"
found_docs = qdrant.similarity_search(query)
```
