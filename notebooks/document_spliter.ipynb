```python
import pandas as pd
```

```python
df = pd.read_csv("./heum_target_urls_20260314011945.csv")
```

```python
from crawl4ai import AsyncWebCrawler
import asyncio

async def save_pages_to_md(urls, output_dir="./md_pages"):
    import os
    os.makedirs(output_dir, exist_ok=True)
    crawler = AsyncWebCrawler()
    for url in urls:
        try:
            # 페이지 크롤링
            result = await crawler.crawl(url)
            content = result['content'] if 'content' in result else ''
            if not content:
                continue
            # 파일명 안전하게
            import re
            basename = re.sub(r"[^a-zA-Z0-9]", "_", url)
            filename = f"{output_dir}/{basename}.md"
            with open(filename, "w", encoding="utf-8") as f:
                f.write(f"# {url}\n\n")
                f.write(content)
            print(f"✅ Saved: {filename}")
        except Exception as e:
            print(f"❌ Error for {url}: {e}")

urls = df['url'].tolist()
asyncio.run(save_pages_to_md(urls))
```

```python
from langchain.document_loaders import WebBaseLoader

# DataFrame 내 URL 리스트로부터 langchain 문서 로딩 예시
urls = df['url'].tolist()
loaders = [WebBaseLoader(url) for url in urls]
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 예시: 한 URL에 대한 텍스트가 있다면(여기서는 url 문자열 자체를 예시로 사용)
# 실제로는 해당 URL에서 가져온 본문 텍스트를 split해야 함
splitter = RecursiveCharacterTextSplitter(chunk_size=20, chunk_overlap=0)

splitted_urls = []
for url in df["url"]:
    # 아래는 예시로 url 문자열 자체를 split하지만,
    # 본문의 텍스트로 대체하여 사용하세요!
    splits = splitter.split_text(str(url))
    splitted_urls.append(splits)

# DataFrame에 결과 컬럼 예시 추가
df["splitted"] = splitted_urls
df[["url", "splitted"]]
```
